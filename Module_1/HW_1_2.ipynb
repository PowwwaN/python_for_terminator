# Робота з пропущеними значеннями

Найпростіше емпіричне правило для опрацювання пропущених значень: якщо пропущених значень більше за 40% - видаляємо колонку, якщо менше за 40% - аналізуємо, як можна заповнити і чи треба.

Зазвичай, якщо відсоток пропущених даних більший за 10 і немає чіткої (яка значно виділяється) моди (найчастішого) значення в даних, то лишаємо дані як є до використання методів машинного навчання. Далі якщо метод вимагає заповення пропущених значень, можемо експериментувати із заповеннями. Також хорошою практикою є створити окрему колонку-флаг (0/1), яка вказує, де були пропущені значення - це буде додаткова ознака для моделі.

# ДЗ 2. Аналіз і заповнення пустих значень у колонках з малим відсотком пропущених

За прикладом вище проведіть аналіз пропущених значень в колонках
- EXT_SOURCE_2
- AMT_GOODS_PRICE

Для швидшого аналізу рекомендую написати фукнцію, яку Ви зможете викликати для кожної з колонок.

Зробіть висновок, що робити з пропущеними значеннями в кожному випадку і виконанайте ту дію, яку зазначили.

```python
def analyze0(x):
    print(null_df_under1[null_df_under1.column_name == x])
    print(df[x].value_counts().head(10))
    print(f"Mode: {df[x].mode()[0]}")
    print(f"Mean: {df[x].mean()}")
    print(f"Median: {df[x].median()}")
    print(df[x].quantile(q = [0.25,0.5,0.75,1]))
    sns.boxplot(df[x])
    plt.show()
    
analyze0('EXT_SOURCE_2')
analyze0('AMT_GOODS_PRICE')
# percentage of empty cells is so negligible(0.21% and 0.09% for EXT_SOURCE_2 and AMT_GOODS_PRICE respectively) 
# that it wouldn't matter much no matter what we fill those cells with
# I would go with median or mode for both columns as it's a straightforward solution.
# In EXT_SOURCE_2 I would go with median rather than mode, 
# as mode isn't that frequent, and isn't close to any of the percentiles.
# AMT_GOODS_PRICE has many outliers so mean doesn't work for sure,
# mode and median happen to have the same value so I would go with it
```

```python
df['EXT_SOURCE_2'] = df['EXT_SOURCE_2'].fillna(df['EXT_SOURCE_2'].median())
df['AMT_GOODS_PRICE'] = df['AMT_GOODS_PRICE'].fillna(df['AMT_GOODS_PRICE'].median())

# confirmation
print(df[['EXT_SOURCE_2', 'AMT_GOODS_PRICE']].isnull().sum())   # both 0
```

### Одномірний (категоріальний vs цільовий) та двомірний (категоріальний vs категоріальний) аналіз (гістограма) - категоріальні стовпчики

```python
import matplotlib.pyplot as plt


def bi_cat_countplot(df, column, hue_column):
    unique_hue_values = df[hue_column].unique()
    fig, axes = plt.subplots(nrows=1, ncols=2)
    fig.set_size_inches(14,6)

    pltname = f'Нормалізований розподіл значень за категорією: {column}'
    proportions = df.groupby(hue_column)[column].value_counts(normalize=True)
    proportions = (proportions*100).round(2)
    ax = proportions.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[0], title=pltname)

    # анотація значень в барплоті
    for container in ax.containers:
        ax.bar_label(container, fmt='{:,.1f}%')

    pltname = f'Кількість даних за категорією: {column}'
    counts = df.groupby(hue_column)[column].value_counts()
    ax = counts.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[1], title=pltname)

    for container in ax.containers:
      ax.bar_label(container)


def uni_cat_target_compare(df, column):
    bi_cat_countplot(df, column, hue_column='TARGET' )


def bi_countplot_target(df0, df1, column, hue_column):
  pltname = 'Клієнт зі складнощами щодо платності'
  print(pltname.upper())
  bi_cat_countplot(df1, column, hue_column)
  plt.show()

  pltname = 'Клієнти зі своєчасними платежами'
  print(pltname.upper())
  bi_cat_countplot(df0, column, hue_column)
  plt.show()
```

### Обчислення мінімального та максимального діапазону відхилень для числових стовпців

```python
def outlier_range(dataset,column):
    Q1 = dataset[column].quantile(0.25)
    Q3 = dataset[column].quantile(0.75)
    IQR = Q3 - Q1
    Min_value = (Q1 - 1.5 * IQR)
    Max_value = (Q3 + 1.5 * IQR)
    return Max_value
```

```python
def analyzeNum(df_num0, df_num1, column_num):
    #print(df_num0[column_num].describe(percentiles = [0.25, 0.5, 0.75, 0.99, 0.999]))

    Q1_num0 = df_num0[column_num].quantile(0.25)
    Q3_num0 = df_num0[column_num].quantile(0.75)
    IQR_num0 = Q3_num0 - Q1_num0
    Min0 = Q1_num0 - 1.5 * IQR_num0
    Max0 = Q3_num0 + 1.5 * IQR_num0

    Q1_num1 = df_num1[column_num].quantile(0.25)
    Q3_num1 = df_num1[column_num].quantile(0.75)
    IQR_num1 = Q3_num1 - Q1_num1
    Min1 = Q1_num1 - 1.5 * IQR_num1
    Max1 = Q3_num1 + 1.5 * IQR_num1

    plt.figure(figsize = (14,6))
    sns.kdeplot(df_num1[df_num1[column_num].between(Min1, Max1)][column_num],label = 'Payment difficulties')
    sns.kdeplot(df_num0[df_num0[column_num].between(Min0, Max0)][column_num],label = 'On-Time Payments')
    plt.ticklabel_format(style='plain', axis='x')
    plt.xticks(rotation = 45)
    plt.legend()
    plt.show()

analyzeNum(df0, df1, 'YEARS_BIRTH') 
# visible spike in payment difficulties in mid 20-ies to late 30-ies.
# increase in on-time payments in mid 50-ies, holds till max age
# (I've only considered clearly large differences, ~3+ percentage points)

analyzeNum(df0, df1, 'AMT_GOODS_PRICE')
# payment difficulties are more common for goods priced ~300k-500k (peak at ~400k-500k).
# on-time payers are slightly more common above ~600k-700k, on-time payer spikes every ~200k after that;
# elsewhere the distributions are almost the same.

analyzeNum(df0, df1, 'DAYS_EMPLOYED')
# the column has a placeholder value 365243 (~18% of rows), removed as an outlier.
# payment difficulties are more common with short employment (up to ~5 years, peak at 1-2 years),
# on-time payers are more common after ~7 years (x axis is negative days: ~-2500 and further left).
```

```python
def analyze_corr(column_x, column_y):
    plt.figure(figsize=(14, 6))
    for i, (data, title) in enumerate([(df1, 'Payment difficulties'), (df0, 'On-Time Payments')], start=1):
        # remove outliers in both columns, keeping the same rows for x and y
        clean = data[(data[column_x] < outlier_range(data, column_x)) &
                     (data[column_y] < outlier_range(data, column_y))]
        corr_all = data[column_x].corr(data[column_y])
        corr_clean = clean[column_x].corr(clean[column_y])
        print(f"{title}: correlation with outliers = {corr_all:.2f}, without = {corr_clean:.2f}")

        plt.subplot(1, 2, i)
        plt.title(f'{title} (corr = {corr_clean:.2f})')
        sns.scatterplot(x=clean[column_x], y=clean[column_y])
        plt.ticklabel_format(style='plain', axis='x')
        plt.ticklabel_format(style='plain', axis='y')
    plt.tight_layout(pad=4)
    plt.show()

analyze_corr('AMT_ANNUITY', 'AMT_CREDIT')
# strong positive correlation (~0.74 and ~0.77): the bigger the credit, the bigger the annuity.
# outliers barely change it, and both groups look alike, so it doesn't explain payment difficulties.
# correlation is about equally strong in both groups, but the upper-right part (large credit, large annuity)
# is less populated for clients with payment difficulties (~5% vs ~9% of the group).
# *could be due to much smaller sample*
```

## Неперервні та категоріальні змінні

```python
def draw_boxplot(df, categorical, continuous, max_continuous, title, hue_column, subplot_position):
    """
    Малює блок-діаграму для заданого DataFrame, категоріальної та неперервної змінної.
    """
    plt.subplot(1, 2, subplot_position)
    plt.title(title)
    red_diamond = dict(markerfacecolor='r', marker='D')
    sns.boxplot(x=categorical,
                y=df[df[continuous] < max_continuous][continuous],
                data=df,
                flierprops=red_diamond,
                order=sorted(df[categorical].unique(), reverse=True),
                hue=hue_column, hue_order=sorted(df[hue_column].unique(), reverse=True))
    plt.ticklabel_format(style='plain', axis='y')
    plt.xticks(rotation=90)

def bi_boxplot(categorical, continuous, max_continuous1, max_continuous0, hue_column):
    """
    Створює паралельні блок-діаграми для двох груп, визначених у наборі даних, на основі
    категоріальної та неперервної змінної, виділяючи відмінності за допомогою відтінків.
    """
    plt.figure(figsize=(16, 10))

    # Графік для першо групи "Труднощі з платежами" (Payment Difficulties)
    draw_boxplot(df1, categorical, continuous, max_continuous1, 'Payment Difficulties', hue_column, 1)

    # Графік для другої групи "Вчасні оплати" (On-Time Payments)
    draw_boxplot(df0, categorical, continuous, max_continuous0, 'On-Time Payments', hue_column, 2)

    plt.tight_layout(pad=4)
    plt.show()
```
